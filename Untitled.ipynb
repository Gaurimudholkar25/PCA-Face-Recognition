# Face Recognition using PCA (Eigenfaces)

## Dataset
This project uses the AT&T Database of Faces (ORL Database).
The dataset will be provided by the course instructor.
Before running the code, upload the provided zip file and rename it to `attface.zip`.

## How to Run
1. Upload `attface.zip` to your Colab session
2. Run all cells in order (Shift + Enter)

## Dependencies
All required libraries are pre-installed in Google Colab except opencv-python.
Run the first cell to install it.

```python
import zipfile
import cv2
import numpy as np
import matplotlib.pyplot as plt
from sklearn.decomposition import PCA

# ── 1. Load face images from zip ──────────────────────────────────────────────
faces = {}
with zipfile.ZipFile("attface.zip") as facezip:
    for filename in facezip.namelist():
        if not filename.endswith(".pgm"):
            continue
        with facezip.open(filename) as image:
            faces[filename] = cv2.imdecode(
                np.frombuffer(image.read(), np.uint8), cv2.IMREAD_GRAYSCALE
            )

# ── 2. Explore the dataset ────────────────────────────────────────────────────
fig, axes = plt.subplots(4, 4, sharex=True, sharey=True, figsize=(8, 10))
faceimages = list(faces.values())[-16:]
for i in range(16):
    axes[i % 4][i // 4].imshow(faceimages[i], cmap="gray")
plt.suptitle("Sample Face Images")
plt.tight_layout()
plt.show()

faceshape = list(faces.values())[0].shape
print("Face image shape:", faceshape)

classes = set(filename.split("/")[0] for filename in faces.keys())
print("Number of classes:", len(classes))
print("Number of images:", len(faces))

# ── 3. Build training matrix (exclude class s40 and s39/10.pgm as test) ───────
facematrix = []
facelabel  = []
for key, val in faces.items():
    if key.startswith("s40/"):
        continue                  # held-out: unknown person
    if key == "s39/10.pgm":
        continue                  # held-out: known person, unseen image
    facematrix.append(val.flatten())
    facelabel.append(key.split("/")[0])

facematrix = np.array(facematrix)
print("Training matrix shape:", facematrix.shape)

# ── 4. PCA → Eigenfaces ────────────────────────────────────────────────────────
pca = PCA().fit(facematrix)

print("\nTop-10 explained variance ratios:")
print(pca.explained_variance_ratio_[:10])

n_components = 50
eigenfaces = pca.components_[:n_components]   # shape: (50, L)

# Visualise first 16 eigenfaces
fig, axes = plt.subplots(4, 4, sharex=True, sharey=True, figsize=(8, 10))
for i in range(16):
    axes[i % 4][i // 4].imshow(eigenfaces[i].reshape(faceshape), cmap="gray")
plt.suptitle("First 16 Eigenfaces")
plt.tight_layout()
plt.show()

# ── 5. Compute weight vectors for every training image ────────────────────────
# weights shape: (n_components, n_train_images)
weights = eigenfaces @ (facematrix - pca.mean_).T
print("Weight matrix shape:", weights.shape)

# ── 6. Helper: recognition function ──────────────────────────────────────────
def recognize(query_img):
    """Return (best_label, best_index, euclidean_distance_array)."""
    q = query_img.reshape(1, -1)
    q_weight = eigenfaces @ (q - pca.mean_).T
    distances = np.linalg.norm(weights - q_weight, axis=0)
    best_idx  = np.argmin(distances)
    return facelabel[best_idx], best_idx, distances

# ── 7. Test A – known person, unseen image (should match s39) ─────────────────
query_img = faces["s39/10.pgm"]
label, idx, dists = recognize(query_img)
print(f"\n[Test A] Best match: {label}  |  Euclidean distance: {dists[idx]:.4f}")

fig, axes = plt.subplots(1, 2, sharex=True, sharey=True, figsize=(8, 6))
axes[0].imshow(query_img, cmap="gray");          axes[0].set_title("Query (s39/10)")
axes[1].imshow(facematrix[idx].reshape(faceshape), cmap="gray")
axes[1].set_title(f"Best match: {label}")
plt.tight_layout(); plt.show()

# ── 8. Test B – unknown person (class s40, never seen during training) ─────────
query_img2 = faces["s40/1.pgm"]
label2, idx2, dists2 = recognize(query_img2)
print(f"\n[Test B] Best match: {label2}  |  Euclidean distance: {dists2[idx2]:.4f}")

fig, axes = plt.subplots(1, 2, sharex=True, sharey=True, figsize=(8, 6))
axes[0].imshow(query_img2, cmap="gray");         axes[0].set_title("Query (s40/1) – unknown")
axes[1].imshow(facematrix[idx2].reshape(faceshape), cmap="gray")
axes[1].set_title(f"Best match: {label2}")
plt.tight_layout(); plt.show()

# ── 9. Bonus – mean face & randomly synthesised face ──────────────────────────
fig, axes = plt.subplots(1, 2, sharex=True, sharey=True, figsize=(8, 6))
axes[0].imshow(pca.mean_.reshape(faceshape), cmap="gray"); axes[0].set_title("Mean Face")

random_weights = np.random.randn(n_components) * weights.std()
newface = random_weights @ eigenfaces + pca.mean_
axes[1].imshow(newface.reshape(faceshape), cmap="gray");   axes[1].set_title("Random Synthetic Face")
plt.tight_layout(); plt.show()
```
